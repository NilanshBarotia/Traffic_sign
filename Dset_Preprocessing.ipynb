```python
merge_pairs = {
    3 : 18,
    4 : 19,
    30 : 36
}
root_path = Path(r'D:\Traffic\traffic_Data_processed\DATA')
for a, b in merge_pairs.items():
    source = root_path/str(b)
    destination = root_path/str(a)

    for img in source.iterdir():
        shutil.move(img, destination/img.name)
    source.rmdir()
```

Now we have merged our folders and now all folders have unique labeled images

Now i will rename the folders

```python
images_path = Path(r'D:\Traffic\traffic_Data_processed\DATA')
folders = sorted(
    [folder for folder in images_path.iterdir() if folder.is_dir()],
    key=lambda folder: int(folder.name)
)
for i, folder in enumerate(folders):
    folder.rename(images_path / f"temp_{i}")
temp_folders = sorted(
    [folder for folder in images_path.iterdir() if folder.name.startswith("temp_")],
    key=lambda folder: int(folder.name.split("_")[1])
)
for i, folder in enumerate(temp_folders):
    folder.rename(images_path / str(i))
```
