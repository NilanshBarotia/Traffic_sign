This file is for deciding which optimier to use

From previous files, the lr for head is 0.0017378008365631102 and lr range for whole model is ((1.9054607491852948e-06)/10, (1.9054607491852948e-06)/4)
The no. of epochs for head remains const. at 3 and no. of epochs for whole model is 8 as found in prev. file
The batch size found is 16 from prev. file

I will ne trying SGD, Adam, AdamW, and RMSProp

```python
import pandas as pd
import regex as re
from fastai.vision.all import *
```

```python
df = pd.read_csv(r'D:\Traffic\labels_processed.csv')
```

```python
def label_function(dpath):
    class_name = re.findall(r'(\d+)_.*\.png$', dpath.name)
    class_id = int(class_name[0])
    return df['Names'][class_id]
```

```python
signs = DataBlock(blocks = (ImageBlock, CategoryBlock),
                  get_items = get_image_files,
                  splitter = RandomSplitter(seed = 42),
                  get_y = label_function,
                  item_tfms = Resize(224),
                  batch_tfms = aug_transforms(size = 224, min_scale = 0.75, do_flip = False))
```

```python
path = Path(r'D:\Traffic\traffic_Data_processed\DATA')
```

```python
lr_head = 0.0017378008365631102
lr_whole_model = 1.9054607491852948e-06
```

```python
set_seed(42, reproducible = True)
```

```python
dls_5_1 = signs.dataloaders(path, bs = 16)
```

```python
model_5_1 = vision_learner(dls_5_1, arch = resnet34, metrics = error_rate, opt_func = SGD)
model_5_1.fit_one_cycle(3, lr_head, wd = 1e-3)
model_5_1.unfreeze()
model_5_1.fit_one_cycle(8, lr_max = slice(lr_whole_model/10, lr_whole_model/4), wd = 1e-3)
```

```python
set_seed(42, reproducible = True)
```

```python
dls_5_2 = signs.dataloaders(path, bs = 16)
```

```python
model_5_2 = vision_learner(dls_5_2, arch = resnet34, metrics = error_rate, opt_func = RMSProp)
model_5_2.fit_one_cycle(3, lr_head, wd = 1e-3)
model_5_2.unfreeze()
model_5_2.fit_one_cycle(8, lr_max = slice(lr_whole_model/10, lr_whole_model/4), wd = 1e-3)
```

```python
set_seed(42, reproducible = True)
```

```python
dls_5_3 = signs.dataloaders(path, bs = 16)
```

```python
model_5_3 = vision_learner(dls_5_3, arch = resnet34, metrics = error_rate, opt_func = Adam)
model_5_3.fit_one_cycle(3, lr_head, wd = 1e-3)
model_5_3.unfreeze()
model_5_3.fit_one_cycle(8, lr_max = slice(lr_whole_model/10, lr_whole_model/4), wd = 1e-3)
```

As we can see that the RMSProp optimizer has worked the best so we will choose that but before finalizing the optimizer we should re run an experiment to have a confirmation
